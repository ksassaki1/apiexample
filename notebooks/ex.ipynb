Introdução
O objetivo é demonstrar como fazer upload de um CSV para o endpoint /upload-csv usando Python, e depois exibir as respostas em um DataFrame do pandas. É necessário ter o servidor rodando em localhost:8000 para que as requisições funcionem.

Logo em seguida, importa requests, pandas, matplotlib.pyplot e io, 
e cria uma pequena função para exibir DataFrame de maneira agradável 
(ou apenas use display(df)).

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

def exibir_resposta_json_em_df(json_data):
    # Converte a parte de agregados_por_categoria em DataFrame
    df_cat = pd.DataFrame(json_data["agregados_por_categoria"])
    # Cria um DataFrame para estatísticas gerais
    df_geral = pd.DataFrame([{
        "total_geral": json_data["total_geral"],
        "media_geral": json_data["media_geral"],
        "contagem_geral": json_data["contagem_geral"]
    }])
    return df_geral, df_cat
```

Dentro de notebooks/, há uma cópia de app/data/sample_data.csv, ou crie um próprio (caso ainda não tenha).

```python
import pandas as pd

df_exemplo = pd.read_csv("../app/data/sample_data.csv")
display(df_exemplo.head())
```

Enviando o CSV para a API
Para enviar o CSV, é preciso abrir o arquivo em modo binário e usar a rota /upload-csv. Em seguida, faça um bloco de código assim:

```python
url = "http://localhost:8000/upload-csv"
caminho_csv = "../app/data/sample_data.csv"

with open(caminho_csv, "rb") as f:
    files = {"file": ("sample_data.csv", f, "text/csv")}
    response = requests.post(url, files=files)

if response.status_code == 200:
    resultado = response.json()
    print("Resposta da API:", resultado)
else:
    print(f"Erro {response.status_code}: {response.text}")
```

Convertendo a resposta em DataFrames e visualizando
Após obter resultado = response.json(), use a função auxiliar para transformar em DataFrames e, em seguida, exibir e até criar um gráfico simples de barras para soma por categoria

```python
df_geral, df_cat = exibir_resposta_json_em_df(resultado)
print("Estatísticas gerais:")
display(df_geral)

print("Agregados por categoria:")
display(df_cat)

# Gráfico de soma por categoria
plt.figure(figsize=(8, 5))
plt.bar(df_cat["categoria"], df_cat["sum"])
plt.title("Soma de valores por categoria")
plt.xlabel("Categoria")
plt.ylabel("Soma de valor")
plt.show()
```

df_cat["sum"] corresponde à soma de valor para cada categoria, conforme retornado pela API, e que o gráfico ilustra essas somas em barras, facilitando a interpretação.

se o usuário enviar outro tipo de arquivo ou um CSV sem as colunas corretas, a API retornará uma mensagem de erro ou um JSON informando que as colunas necessárias não foram encontradas.
